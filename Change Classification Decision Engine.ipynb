```python
#Creating Model Change Register
change_register=df_clean.copy()
#Unique Model IDs
change_register.insert(
    0,"Change_ID",["MC"+str(i).zfill(4) for i in range(1, len(change_register)+1)]
)

change_register.head()
```

```python
#Creating Model Identifier
change_register["Model_ID"]="CR_PD_001"
change_register["Model_Type"]="Credit Risk-PD Model"
change_register[["Change_ID","Model_ID","Model_Type"]].head()
```

```python
# Assign different model types to the change requests

model_ids = [
    "CR_PD_001",
    "CR_LGD_001",
    "AML_TM_001",
    "FRD_001"
]

model_types = [
    "Credit Risk - Probability of Default",
    "Credit Risk - Loss Given Default",
    "AML - Transaction Monitoring",
    "Fraud Detection"
]

np.random.seed(42)

change_register["Model_ID"] = np.random.choice(
    model_ids,
    size=len(change_register)
)

model_mapping = dict(zip(model_ids, model_types))

change_register["Model_Type"] = change_register["Model_ID"].map(model_mapping)

change_register[[
    "Change_ID",
    "Model_ID",
    "Model_Type"
]].head(10)
```

```python
change_register["Model_Type"].value_counts()
```

```python
#simulate type of change the governance team received.
change_types = [
    "Parameter Change",
    "Data Change",
    "Methodology Change",
    "Regulatory Change",
    "Implementation Change",
    "Documentation Change"
]

np.random.seed(42)

change_register["Change_Type"] = np.random.choice(
    change_types,
    size=len(change_register),
    p=[0.25, 0.20, 0.15, 0.10, 0.15, 0.15]
)

change_register["Change_Type"].value_counts()
```

```python
#Creating Materiality Class
change_register["Materiality"] = pd.cut(
    change_register["credit_amount"],
    bins=[0, 2000, 8000, float("inf")],
    labels=["Low", "Medium", "High"]
)

change_register[
    ["Change_ID", "credit_amount", "Materiality"]
].head(10)
```

```python
change_register["Materiality"].value_counts()
```

```python
#Creating Regulatory Trigger
change_register["Regulatory_Trigger"] = np.where(
    change_register["Change_Type"] == "Regulatory Change",
    "Yes",
    "No"
)

change_register["Regulatory_Trigger"].value_counts()
```

```python
#Create Methodology_Change
change_register["Methodology_Change"]=np.where(change_register["Change_Type"]=="Methodology Change","Yes","No")
change_register["Methodology_Change"].value_counts()
```

```python
#Create Data Change(None,Minor,Major)
change_register["Data_Change"] = np.select(
    [change_register["Change_Type"] == "Data Change",
        change_register["Change_Type"] == "Methodology Change"],
    [np.where(change_register["credit_amount"] > 8000,"Major","Minor"),"Minor"],
    default="None")

change_register["Data_Change"].value_counts()
```

```python
#Create Validation_Required
change_register["Validation_Required"] = np.where(
    ((change_register["Materiality"] == "High") |(change_register["Methodology_Change"] == "Yes") |
        (change_register["Regulatory_Trigger"] == "Yes")),
    "Yes",
    "No")

change_register["Validation_Required"].value_counts()
```

```python
#Create Production Impact- using loan duration as proxy for complexity/exposure
change_register["Production_Impact"] = pd.cut(
    change_register["duration_months"],
    bins=[0, 12, 24, float("inf")],
    labels=["Low", "Medium", "High"]
)

change_register["Production_Impact"].value_counts()
```

```python
#Create Previous_Issues as simulation proxy to see underlying population model has problematic outcomes.

change_register["Previous_Issues"] = np.where(
    change_register["credit_risk"] == 0,
    "Yes",
    "No")

change_register["Previous_Issues"].value_counts()
```

```python
change_register[
    [
        "Change_ID",
        "Model_ID",
        "Model_Type",
        "Change_Type",
        "Materiality",
        "Regulatory_Trigger",
        "Methodology_Change",
        "Data_Change",
        "Validation_Required",
        "Production_Impact",
        "Previous_Issues"
    ]
].head(20)
```

```python
#Create a Scoring Engine
def calculate_risk_score(row):

    score = 0

    # Materiality
    if row["Materiality"] == "High":
        score += 30
    elif row["Materiality"] == "Medium":
        score += 15

    # Regulatory trigger
    if row["Regulatory_Trigger"] == "Yes":
        score += 30

    # Methodology change
    if row["Methodology_Change"] == "Yes":
        score += 25

    # Data change
    if row["Data_Change"] == "Major":
        score += 20
    elif row["Data_Change"] == "Minor":
        score += 10

    # Validation
    if row["Validation_Required"] == "Yes":
        score += 15

    # Production impact
    if row["Production_Impact"] == "High":
        score += 20
    elif row["Production_Impact"] == "Medium":
        score += 10

    # Previous issues
    if row["Previous_Issues"] == "Yes":
        score += 10

    return score
```

```python
change_register["Risk_Score"] = change_register.apply(
    calculate_risk_score,
    axis=1)
```

```python
change_register[
    ["Change_ID", "Change_Type", "Materiality", "Risk_Score"]].head(20)
```

```python
#Using the above risk score for governance classification
def classify_change(score):

    if score >= 70:
        return "Major"
    elif score >= 35:
        return "Moderate"
    else:
        return "Minor"

change_register["Classification"] = change_register["Risk_Score"].apply(
    classify_change
)
```

```python
change_register["Classification"].value_counts()
```

```python
def governance_action(classification):

    if classification == "Major":
        return "Model Risk Committee Review"

    elif classification == "Moderate":
        return "Independent Validation"

    else:
        return "Governance Team Approval"

change_register["Required_Action"] = change_register["Classification"].apply(
    governance_action
)
```

```python
#Checking the completed Scoring Engine
change_register[
    [
        "Change_ID",
        "Model_ID",
        "Change_Type",
        "Materiality",
        "Regulatory_Trigger",
        "Methodology_Change",
        "Data_Change",
        "Validation_Required",
        "Production_Impact",
        "Risk_Score",
        "Classification",
        "Required_Action"
    ]
].head(20)
```

```python
#Measuring Business Impact
classification_summary = (
    change_register["Classification"]
    .value_counts()
    .reindex(["Minor", "Moderate", "Major"])
)

classification_summary
```

```python
total_changes = len(change_register)

minor_changes = (
    change_register["Classification"] == "Minor"
).sum()

automation_rate = minor_changes / total_changes

print("Total Change Requests:", total_changes)
print("Automatically Triaged:", minor_changes)
print(f"Automation Rate: {automation_rate:.1%}")
```

```python
#Updating Risk scoring function
def calculate_risk_score(row):

    score = 0

    # 1. Materiality
    if row["Materiality"] == "High":
        score += 30
    elif row["Materiality"] == "Medium":
        score += 15

    # 2. Regulatory trigger
    if row["Regulatory_Trigger"] == "Yes":
        score += 30

    # 3. Methodology change
    if row["Methodology_Change"] == "Yes":
        score += 25

    # 4. Data change
    if row["Data_Change"] == "Major":
        score += 20
    elif row["Data_Change"] == "Minor":
        score += 10

    # 5. Validation requirement
    if row["Validation_Required"] == "Yes":
        score += 15

    # 6. Production impact
    if row["Production_Impact"] == "High":
        score += 20
    elif row["Production_Impact"] == "Medium":
        score += 10

    # 7. Previous model issues
    if row["Previous_Model_Issues"] == "Yes":
        score += 10

    return score
```

```python
change_register["Risk_Score"] = change_register.apply(
    calculate_risk_score,
    axis=1)
```

```python
#Re-classify
change_register["Classification"] = change_register["Risk_Score"].apply(
    classify_change)
```

```python
#Re-calculating automation rate
total_changes = len(change_register)

minor_changes = (
    change_register["Classification"] == "Minor"
).sum()

automation_rate = minor_changes / total_changes

print("Total Change Requests:", total_changes)
print("Automatically Triaged:", minor_changes)
print(f"Automation Rate: {automation_rate:.1%}")
```

```python
#Building the governance decision
def governance_action(classification):

    if classification == "Major":
        return "Model Risk Committee Review"

    elif classification == "Moderate":
        return "Independent Validation"

    else:
        return "Governance Team Approval"

change_register["Required_Action"] = change_register["Classification"].apply(governance_action)
```
